# Práctica — ¿Quién va a abandonar el curso?

Trabajás como analista en **Lumen**, una edtech argentina de cursos online pagos, con cohortes y tutor. Cerca del 20% de los alumnos abandona antes de terminar, y cada abandono es plata perdida y una reseña mala.

El equipo académico puede **intervenir**: un tutor llama al alumno, le ofrece una tutoría extra o pausar la cursada. Pero los tutores no alcanzan para llamar a todos. Tu jefa te pide:

> *"Al terminar la **semana 2** de cada curso, quiero una lista de los alumnos con más riesgo de abandonar, para que los tutores sepan a quién llamar primero."*

**Con qué contás:** `abandono_cursos.csv` — las inscripciones de las últimas cohortes, con cómo terminó cada una.

**La regla de la práctica:** cada cosa que afirmes, mostrala. Todo hallazgo y toda decisión van acompañados de un output o un gráfico, y de una línea que diga qué ves.

**Cómo está organizada:** **A. Los datos** (secciones 2 a 5) → **B. El modelo** (6 y 7) → **C. Los resultados** (8 a 10). Es el mismo orden que vas a seguir en cualquier proyecto, con cualquier dataset.

## 5. ¿Qué variables entran al modelo?

Esta es la decisión más importante del proyecto. Para **cada** columna, hacete estas tres preguntas:

1. **¿Existe en el momento en que voy a usar el modelo?** Releé la consigna de tu jefa: ¿cuándo se predice? Si una variable se conoce recién después, **no se puede usar**, por bien que funcione. Esto se llama *data leakage*.
2. **¿Dice algo que otra variable ya dice?** Si dos columnas cuentan lo mismo, tener las dos no suma y puede ensuciar la interpretación.
3. **¿Aporta algo?** Una variable sin ninguna relación con el target es ruido, y el ruido le da al modelo más oportunidades de memorizar.

**Probalo, no lo supongas:** si sospechás que una variable es tramposa, entrená un modelo rápido con y sin ella y compará. Un resultado demasiado bueno es una señal de alarma, no un logro.

## Para llevarte: el checklist de cualquier proyecto

Estas preguntas no son de este dataset: sirven para **cualquier** problema de machine learning. Guardalas.

**Los datos**
- ¿Qué representa cada fila? ¿Cuál es el target?
- ¿Hay duplicados, nulos, categorías mal escritas, valores imposibles? ¿Los nulos **significan** algo?
- ¿Cómo está repartido el target? ¿Qué acierta alguien que no hace nada?
- ¿Graficaste **todas** las variables contra el target, no solo las que esperabas?
- ¿Hay relaciones que no sean rectas?

**Las variables**
- ¿Cada variable existe **en el momento en que se va a predecir**?
- ¿Hay variables que dicen lo mismo?
- ¿Hay variables que no dicen nada?

**El modelo**
- ¿Separé train y test antes de mirar resultados? ¿Hacía falta estratificar?
- ¿El modelo necesita que escale?
- ¿Le gano al baseline?
- ¿Miré train **y** test?

**Los resultados**
- ¿La métrica que miro es la que le importa al negocio?
- ¿Qué cuesta más, un falso positivo o un falso negativo?
- ¿Mi mejor número puede ser demasiado bueno para ser verdad?
- ¿Puedo explicar el resultado sin nombrar el algoritmo?
